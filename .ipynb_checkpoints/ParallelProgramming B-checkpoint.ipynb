# Programación Paralela

El procesamiento paralelo es una técnica informática en la que coexisten múltiples flujos de cálculos o tareas de procesamiento de datos a través de $\textbf{numerosas unidades centrales de procesamiento (CPU)}$ que trabajan simultáneamente (Estas tareas pueden ejecutarse simultáneamente: en otro núcleo de procesador, en otro procesador o en una computadora completamente separada (sistemas distribuidos)).

# Concurrencia Vs Paralelismo

## Concurrencia: 
concurrencia es la capacidad del CPU para procesar más de un proceso al mismo tiempo, ¿Pero que implica esto realmente?. Pues bien, para comprender esto es necesario entender cómo funciona un procesador. Un procesador puede procesar al mismo tiempo el mismo número de procesos que el número de CORES que tiene, de esta forma, si un procesador tiene un CORE, entonces solo podrá ejecutar un proceso a la vez, por otro parte, si tenemos 8 CORES, entonces podremos ejecutar hasta 8 procesos al mismo tiempo

## Paralelismo:
El paralelismo sigue la filosofía de “divide y vencerás”, ya que consiste en tomar un único problema, y mediante concurrencia llegar a una solución más rápido. El paralelismo lo que hace es tomar el problema inicial, dividir el problema en fracciones más pequeñas, y luego cada fracción es procesada de forma concurrente, aprovechando al máximo la capacidad del procesador para resolver el problema.

La $\textbf{principal diferencia}$ del paralelismo contra la concurrencia es que, en el paralelismo, todos los procesos concurrentes están íntimamente relacionados a resolver el mismo problema, de tal forma que el resultado de los demás procesos afecta al resultado final, en la concurrencia (por tanto no importan los resultados individuales), no importa que los procesos esten relacionados, solo importa que se realizan al mismo tiempo.

 

imagenes tomadas de [1]

 

## Ejemplo de concurrencia:
Empecemos con la concurrencia, imagina una aplicación de descarga de música, en la cual puedes descargar un número determinado de canciones al mismo tiempo, cada canción es independiente de la otra, por lo que la velocidad y el tiempo que tarde en descargarse cada una no afectara al resto de canciones. Esto lo podemos ver como un proceso concurrente, ya que cada descarga es un proceso totalmente independiente del resto





## Ejemplo de paralelismo

En el caso del paralelismo podemos imaginas la clásica página de viajes, donde nos ayudan a buscar el vuelo más barato o las mejores promociones, para hacer esto, la página debe de buscar al momento en cada aerolínea el vuelo más barato, con menos conexiones, etc. Para esto puedo hacerlo de dos formas, buscar secuencialmente en cada aerolínea las mejores promociones (muy tardado) o utilizar el paralelismo para buscar al mismo tiempo las mejores promociones en todas las aerolíneas




Observemos en la imagen como el proceso parte de una entrada inicial (inputs) los cuales definen las características del vuelo a buscar, luego se utiliza la concurrencia para buscar en las cuatro aerolíneas al mismo tiempo. Veamos que en este proceso es indispensable que las 4 búsquedas terminen para poder arrojar un resultado. Podemos ver claramente la relación entre los 4 procesos, ya que el resultado de uno puedo afectar al proceso final.

### Sistemas de memoria distribuida
En un multiprocesador de memoria distribuida cada procesador tiene una memoria local y ejecuta su propio programa de nodo. El programa puede alterar valores en la memoria local del procesador en ejecución y puede enviar datos en forma de mensajes a los otros procesadores de la red. La interconexión de los procesadores define la topología de la red (se le dice topología, pues no nos importa donde están las cosas o que ttan grandes son, lo que nos preocupa es como estan conectadas. Por ejemplo en un grafo se puede definir una topologia aparir de la matrica definida por las conexiones entre los nodos) y un ejemplo simple que es suficiente para nuestra introducción es el anillo. (imagen tomada de [3])







cada procesador tiene un numero de identificación. Decimos que el procesador $Proc(\lambda)$ esta en la vecinda del procesador $Proc(\mu)$ si hay una conección derecta entre ellos


$\textbf{Comunciación}$

se utiliza la siguiente notación para indicar la comunciación entre procesadores



En este modelo si $Proc(\mu)$ ejecuta la instrucción $send(V_{loc},\lambda)$, entonces una copia de la matriz local $V_{loc}$ es enviada a $Proc(\lambda)$ y la ejucución del nodo $Proc(\mu)$ se reanuda inmediatamente, es permitido para un procesador enviarse mensages a si mismo. Para enfatizar que la matriz esta guarda en la memoria local se utiliza, el sub indice "$loc$".

Si $Proc(\mu)$ ejecuta la instrucción $recv(U_{loc},\lambda)$ luego la ejecución de su programa de nodo se suspende hasta que se recive un mensaje de $Proc(\lambda)$. Una vez recibido el mensaje se coloca en una matriz local $U_{loc}$ y $Proc(\mu)$ reanuda la ejecución de su programa de nodo

Para tener en cuenta, en la práctica, un mensaje es una cadena de bits y se debe proporcionar un encabezado que indique al receptor las dimensiones de la matriz y el formato de las palabras en coma flotante que se utilizan para representar sus entradas. Pasar del masaje a la matriz almacenada lleva tiempo, pero es un gasto general que no intentamos cuantificar.

$\textbf{Alguna estructura de datos distribuidos.}$

Antes de que podamos especificar nuestro primer algoritmo de memoria distribuida, debemos considerar la cuestión del diseño de los datos. ¿Cómo se distribuyen las matrices y los vectores participantes en la red?.

Suponga que $x\in \mathbb{R}^n$ va a ser distribuido entre las memorias locales de una red de $p$- procesadores. Asuma por el momento que $n=rp$. Dos aproximaciones canonicas a este problema son guardado por filas y guardado por columnas.

$\textbf{Por columnas}$:

En el guradado por columnas nosotros consideramos un vector $x$ como una matriz $r\times p$.


$$x_{r\times p}=\begin{bmatrix} x(1:r) & x(r+1:2r) & \dots  & x((p-1)r+1: rp)  \end{bmatrix}$$

y cada columna se guarda en un procesador, es decir $x((\mu-1)r+1: \mu r)\in Proc(\mu)$, note que cada procesador tiene una porcion contigua de $x$

$\textbf{Por filas}$:

En el guardado fila por comlumna nosotros consideramos a $x$ como una $p\times r$ matriz



$$x_{p\times r}=\begin{bmatrix} x(1:p) & x(p+1:2p) & \dots  & x((r-1)p+1: rp)  \end{bmatrix}$$

y se guarda cada fila en el procesador correspondiente, es decir, $x(\mu: p: n)\in Proc(\mu)$ ($x(\mu: p: n)$ significa paraticionar $x$, en $p:n$ partes y selecionar de cada parte la entrada $\mu$ por ejemplo $x(2: 4:20)=\begin{bmatrix} x_2 &x_6 & x_ {10} & x_{14} & x_{18} \end{bmatrix}$ $x$ se partio en 5 partes cada uno con 4 elementos, y se selecciono el segundo de cada parte), a este método se le refiere como el método envolvente para distruir un vector porque los componentes de $x$ pueden considerarse como cartas en una baraja que see "reparten" a los procesadores de forma envolvente.


Si $n$ no es un multiplo exacto dee $p$, entonces estas mismas ideas se pueden aplicar con una pequeña modificación, por ejemplo considere el guardado por columna con $n=14$ y $p=4$

$$x^T=\begin{bmatrix}x_1  x_2 x_3x_4 & | & x_5x_6x_7x_8 &|&x_9x_{10}x_{11} &|&x_{12}x_{13}x_{14}\end{bmatrix}$$

cada una de las columnas anteriores se guarda en el procesador 1,2,3 y 4 respectivamente. En general si $n=pr+q$ con $0\leq q < p$, entonces $Proc(1),\dots,Proc(q)$, puede tener  cada uno $r+1$ componentes y el resto $Proc(q+1),\dots,Proc(p)$ tiene $r$ componenetes. En el guradado por filas es analogo, se coloca una ultima columna con $q$ componentes, y por ende las primeras $q$ filas tendrian $r+1$ componentes y las siguientes filas $r$ componentes.

Similarmente si tenemos una matriz $A\in \mathbb{R}^{n\times n}$ y $n=rp$

# Gaxpy en un Anillo

## Multiplicación

Ahora vamos a desarrollar un algoritmo en un anillo de procesadores, para gaxpy $z=y+Ax$ ($A\in \mathbb{R}^{n\times n}, x,y\in\mathbb{R}^n$), para claridad, asuma que $n=rp$, donde $p$ es el tamaño del anillo, se partciona de la siguiente manera el sistma $z=y+Ax$ (La multiplicación de  dos matrices $C=AB$, viendolo como el producto exterior entre las columas de $A$ y las filas de $B$, se conoce como "outer product" o operacion de  Gaxpy:  https://en.wikipedia.org/wiki/Outer_product )



$$\begin{bmatrix}z_1\\\vdots\\z_p\end{bmatrix}=\begin{bmatrix}y_1\\\vdots\\y_p\end{bmatrix}+\begin{bmatrix} A_{11} &\cdots& A_{1p} \\ \vdots & &\vdots\\A_{p1}&\cdots& A_{pp}\end{bmatrix}\begin{bmatrix} x_1 \\\vdots\\ x_p\end{bmatrix}$$

donde $A_{ij}\in \mathbb{R}^{r\times r}$ y $x_i,y_i,z_i\in\mathbb{R}^r$, nosotros asumimos que que el procesador $\mu$ empieza guardando $x_{\mu},y_{\mu}$ y la fila $\mu$ de A. Desde la perpectiva de $Proc(\mu)$ el cálculo de 

$$z_{\mu}=y_{\mu}+ \sum_{r=1}^p A_{\mu r}x_r$$

Involucra la información local $(A_{\mu r}, y_{\mu},x_{\mu})$ y la información no local $x_{\tau}, \tau\neq \mu$. Para que las porciones no locales de $x$ estén disponibles, circulamos sus subvectores alrededor del anillo. Por ejemplo, en el caso $P=3$ rotamos $x_1,x_2$ y $x_3$ como sigue



cuando un subvector de $x$ "visita", el procesador host debe incorporar el término apropiado en su suma corriente, como sigue 








Ahora el alogortimo para calcular $z=y+Ax$

**Algoritmo 6.11:** Suponga que $A\in \mathbb{R}^{n\times n}$, y que $x\in \mathbb{R}^n$, $n=rp$ estan dados, si tenemos un anillo de  $p$-procesadores, entonces se ejecuta el siguiente programa de nodo, Asuma las siguientes inciciaciones de memoria:
$row=1+(\mu-1)r:\mu r$, $A_{loc}=A(row,:)$, $x_{loc}=x(row)$, $y_{loc}=y(row)$



```julia
@everywhere function Mult(mu, dy, dA, x) # Algoritmo 6.1.1
    for t in 1:p
        tau = mu - t - 1
        if tau <= 0  tau += p  end
        
        #dx = distribute(x, dist=(p,1), procs=Ids[t])
        #Ind = circshift(Indices,t)
        #dy[:L] = dy[:L] + dA[:L][:,Ind[p-1]]*dx[:L]
        
        dy[:L] = dy[:L] + dA[:L][:,1+(tau-1)*r:tau*r]*x[1+(tau-1)*r:tau*r] #dx[:L]
    end    
    #println(tau," ",1+(tau-1)*r," ", tau*r, " ", x[1+(tau-1)*r:tau*r])
end

[@spawnat mu Mult(mu, dy, dA, x) for mu in workers()]
[@fetchfrom mu dy[:L] for mu in workers()] # print result

# ------------ utilizando el libro --------------
@everywhere Ids = map(i->circshift(workers(),-i),1:p)
Indices = map(i->dy.indices[i][1],1:p)
#[println(i, " ", mu) for i in 1:p for mu in workers()]
for t=1:p
    dx = distribute(x, dist=(p,1), procs=Ids[t])
    Ind = circshift(Indices,t)
    [@spawnat p dy[:L]=dy[:L]+dA[:L][:,Ind[p-1]]*dx[:L] for p in workers()]
end
[@fetchfrom mu dy[:L] for mu in workers()] # print result
```

# Shared Memory 
## Condicion de Carrera
(https://stackoverflow.com/questions/34510/what-is-a-race-condition)
Una condición de carrera ocurre cuando dos o más subprocesos pueden acceder a datos compartidos e intentan cambiarlos al mismo tiempo. Debido a que el algoritmo de programación de subprocesos puede intercambiar entre subprocesos en cualquier momento, no se sabe el orden en el que los subprocesos intentarán acceder a los datos compartidos. Por lo tanto, el resultado del cambio en los datos depende del algoritmo de programación de subprocesos, es decir, ambos subprocesos "compiten" para acceder/cambiar los datos.

Los problemas ocurren a menudo cuando un hilo hace una "verificación y luego actúa"  y otro hilo hace algo con el valor en entre el "cheque" y el "acto". P.ej:




# Shared Memory Sytems

Todas las preocupaciones relacionadas con el cálculo de la memoria distribuida se encuentran en forma modificada. El procedimiento general debe tener equilibrio de carga y los cálculos deben organizarse de manera que los procedimientos individuales tengan que esperar el menor tiempo posible hasta que se calcule algo útil. El tráfico entre las memorias global y local debe gestionarse con cuidado, porque el alcance de dicha transferencia de datos suele suponer una sobrecarga importante. La naturaleza de la conexión física entre los procesadores y la memoria compartida es muy importante y puede afectar el desarrollo algotímico. sin embargo, por simplicidad consideramos este aspecto del sistema como una caja negra.




$\textbf{A shared memory Gaxpy}$

Conidere la siguiente parteción del problema $z=y+Ax$

$$\begin{bmatrix}z_1\\\vdots\\z_p\end{bmatrix}=\begin{bmatrix}y_1\\\vdots\\y_p\end{bmatrix}+\begin{bmatrix} A_{1}\\ \vdots \\ A_{p} \end{bmatrix}x$$

vamos asumir que $n=rp$ y que $A_{\mu}\in \mathbb{R}^{r\times n}$, $y_{\mu}\in\mathbb{R}^r$ y $z_{\mu}\in \mathbb{R}^r$.



Asumimos que una copia de este programa reside en cada procesador. 

Los datos se transfieren hacia y desde la memoria global durante la ejecución del algoritmo. Hay dos lecturas de memoria global antes del bucle ($x_{loc}$ y $y_{loc}$), una leída cada vez a través del bucle ($a_{loc}=A(row,j)$) y una escritura después del bucle ($y(row)=y_{loc}$). sólo un procesador escribe en una ubicación de memoria global determinada en $y$. por lo que no es necesario sincronizar los procesadores participantes. Cada uno tiene una parte completamente independiente de la operación general de gaxpy y no tiene que monitorear el progreso de los otros procesadores; el cálculo está programado estáticamente porque la partición del trabajo se realiza antes de la ejecución.



```julia
xs = SharedArray{Float64}((n,1), init = S -> S[:]=x[:],      pids=workers())# , pids=workers())#mode="+r", 
ys = SharedArray{Float64}((n,1), init = S -> S[:]=y[:],      pids=workers())
As = SharedArray{Float64}((n,n), init = S -> S[:,:]=dA[:,:], pids=workers())
zs = As*xs+ys;

@everywhere function MultShared(mu, xs, ys, As)# algoritmo 6.1.4 
    yloc = ys[1+(mu-1)*r:mu*r];
    #println("id:",myid()," ",1+(mu-1)*r, " ",mu*r,"\n")
    for j = 1:n
        aloc = As[1+(mu-1)*r:mu*r,j]
        yloc = yloc + aloc*xs[j]
    end
    display(ys)
    ys[1+(mu-1)*r:mu*r] = yloc;
end

#@btime 
[@spawnat mu MultShared(mu, xs, ys, As)  for mu in workers().-1];
sleep(0.1)
ys
#[@fetchfrom mu ys for mu in workers()]  
```

Ahora presentamos un algoritmo, en donde se tienen que sincronizar los procesadores, para que se puede ejecutar la multiplición de matrices exitosamente en un sistema de memoria compartida.



Para entender la barrera, es conveniente considerar un procesador como bloqueado o libre. Un procesador queda bloqueado y suspende la ejecución cuando ejecuta la barrera. Una vez bloqueado el procesador pth, todos los procesadores vuelven al "estado libre" y reanudan la ejecución.

En el $\textbf{Algoritmo 6.1.5}$, el procesador se bloquea después de calcular su porción del producto matriz-vector, no podemos predecir el orden en el que se producirán estos bloqueos, pero una vez que el último procesador alcanza la barrera, todos se liberan y $Proc(1)$ puede realizar la suma de vectores.

```julia
Ws = SharedArray{Float64}((n,p), pids=workers());
@everywhere function MatrixSharedCol(mu, xs, ys, As, Ws) #algoritmo 6.1.5
    Aloc = As[:, 1+(mu-1)*r:mu*r]
    xloc = xs[1+(mu-1)*r:mu*r]
    display(Aloc*xloc)
    Ws[:,mu] = Aloc*xloc
    ###=
    sleep(1.0)
    if mu==nworkers()
        yloc = ys
        for j=1:p
            wloc = Ws[:,j];
            yloc = yloc + wloc;
        end
        ys[:] = yloc; #ones(n);
        #display("Hola mundo")
        display(ys)
    end
    ##=#
end

[@spawnat mu+1 MatrixSharedCol(mu, xs, ys, As, Ws)  for mu in workers().-1];
ys
#sleep(0.1)
#@spawnat 1 MatrixSharedCol(0, xs, ys, As)
```

$\textbf{Programación dinámica}$

En vez de tener un solo procesador haciendo la suma de vectores,Es tentador que cada procesador agregue su contribución directamente a la variable global, sin embargo esto podria ser muy problematico , pues podriamso estar perdiendo las contribuciones de algunos porcesadores, considere la siguiente situación


$Proc(1)$ lee $y$ 

$Proc(2)$ lee $y$ 

$Proc(1)$ escribe  $y$ 

$Proc(2)$ escribe  $y$ 

La contribución de $Proc(1)$ se pierde porque $Proc (1)$ y $Proc(2)$ obtienen la misma version de $y$. como resultado la escritura de  $Proc(1)$ se pierde por la escritura de $Proc(2)$

Para evitar que suceda este tipo de cosas, los sistemas de memoria más populares apoyan la idea de una sección crítica. Estas son partes especiales y aisladas de un programa de nodo que requieren una clave para ingresar. En todo el sistema, solo hay una clave y, por lo tanto, el efecto neto es que solo un procesador puede ejecutarse en una sección crítica en un momento dado.




Este uso del concepto de sección crítica controla la actualización de y de una manera que garantiza la corrección. El algoritmo está programado dinámicamente porque el orden en que ocurren las sumas se determina a medida que se desarrolla el cálculo. La clasificación dinámica es muy importante en problemas con estructura irregular.
