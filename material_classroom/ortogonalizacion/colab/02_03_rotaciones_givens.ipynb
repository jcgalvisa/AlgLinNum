# Rotaciones de Givens

```julia
function Givens(a,b)
    if b!=0
        if abs(b)>abs(a)
            τ=-a/b
            s=-1/sqrt(1+τ^2)
            c=s*τ
        else
            τ=-b/a
            c=1/sqrt(1+τ^2)
            s=c*τ
        end
    end
    return c,s
end

            
```

```julia
function Giv(B, i, k, c, s)
    A=copy(B)
    for j=1:size(A)[2]
        τ1 = A[i,j]
        τ2 = A[k,j]
    
        A[i,j] = c*τ1 - s*τ2
        A[k,j] = s*τ1 + c*τ2
    end
    return A
end
```

```julia
A = floor.(10*rand( 3, 3))
```

```julia
c, s = Givens(A[2,1], A[3,1])
C= Giv(A , 2,3 ,c,s)
```
