# Iteración inversa

Supongamos que el método de potencias se aplica reemplazando $A \in \mathbb{R}^{n\times n}$ simétrica por $(A-\lambda I)^{-1}$. Si $\lambda$ es muy cercano a un eigenvalor distinto de $A$, entonces el siguiente vector iterado va a caer muy cerca en la correspondiente eigendirección.

Para observar esto, recordemos que $A \in \mathbb{R}^{n\times n}$ es simétrica y por ende diagonalizable, por lo que existe $Q=[q_{1}, \cdots , q_{n}]$ ortogonal tal que $$Q^{T}AQ=diag(\lambda_{1},\cdots,\lambda_{n})$$

Notemos que si $x= \sum_{i=1}^{n}a_{i}q_{i}$ y como $Aq_{i} = \lambda_{i}q_{i}$ para $i=1:n$, entonces tenemos que $(A-\lambda I)^{-1}x = y = \sum_{i=1}^{n}b_{i}q_{i}$ y así

\begin{align*}
x = (A-\lambda I)y &= (A-\lambda I)\sum_{i=1}^{n}b_{i}q_{i} \\
&= \sum_{i=1}^{n}b_{i}Aq_{i} - \sum_{i=1}^{n}\lambda b_{i}q_{i} \\
&= \sum_{i=1}^{n}b_{i}\lambda_{i}q_{i} - \sum_{i=1}^{n}\lambda b_{i}q_{i} \\
&= \sum_{i=1}^{n}b_{i}\lambda_{i}q_{i} - \lambda b_{i}q_{i} \\
&= \sum_{i=1}^{n}(\lambda_{i}-\lambda)b_{i}q_{i}
\end{align*}

Por consiguiente, se tiene que $a_{i} = (\lambda_{i}-\lambda)b_{i}$ es decir que $b_{i} = \frac{a_{i}}{\lambda_{i}-\lambda}$. De esta manera, tenemos que $$(A-\lambda I)^{-1}x = \sum_{i=1}^{n}\frac{a_{i}}{\lambda_{i}-\lambda}q_{i}$$

Así, si $\lambda \approx \lambda_{j}$ y $a_{j}$ no es muy pequeño, entonces este vector $y$ que corresponde a la iteración del método de potencias tiene una fuerte componente en la dirección $q_{j}$. Este proceso se llama $\textbf{Iteración inversa}$ y requiere la solución de un sistema lineal cuya matriz de coeficientes es $A-\lambda I$.



# Iteración cociente de Rayleigh

Supongamos que $A \in \mathbb{R}^{n\times n}$ es simétrica y que $x$ es un $n$-vector no nulo dado. Consideremos la función $m(\lambda)=||(A-\lambda I)x||_{2}$, entonces

\begin{align*}
m^{2}(\lambda) &= [(A-\lambda I)x]^{T}(A-\lambda I)x \\
&= x^{T}(A^{T}-\lambda I)(A-\lambda I)x \\
&= x^{T}(A-\lambda I)(A-\lambda I)x \\
&= x^{T}(A^{2}-2\lambda A + \lambda^{2}I)x \\
&= x^{T}A^{2}x - 2\lambda x^{T}Ax + \lambda^{2}x^{T}x
\end{align*}

Por lo tanto, derivando se tiene que

\begin{align*}
2m(\lambda)\cdot m'(\lambda) &= -2x^{T}Ax + 2\lambda x^{T}x
\end{align*}

Así, si $m'(\lambda)= 0$ se sigue que $0 = -2x^{T}Ax + 2\lambda x^{T}x$ y por ende $\lambda = r(x) = \frac{x^{T}Ax}{x^{T}x}$.

De esta manera, diferenciando encontramos que $$\lambda = r(x) = \frac{x^{T}Ax}{x^{T}x}$$ minimiza la función $m(\lambda)=||(A-\lambda I)x||_{2}$. El escalar $r(x)$ se conoce como el $\textbf{cociente de Rayleigh}$ de $x$. Notese que si $x$ es aproximadamente un vector propio, entonces $r(x)$ es una elección razonable para su correspondiente valor propio.

Combinando esta idea con la iteración inversa, podemos realizar la $\textbf{Iteración cociente de Rayleigh}$. La Iteración cociente de Rayleigh casi siempre converge y cuando converge lo hace a una tasa de convergencia cubica.

```julia
function RAY(A,x,n)
    x = x/norm(x,2)
    μ = 0
    for i = 1:n
        μ = (x'*A*x)/(x'*x)
        B = (A-μ*I)
        z = B\x
        x = z/norm(z,2)
        println("El eigenvalor es = \n",μ)
    end
    return μ
end
```

```julia
A=[1 1 1 1 1 1; 1 2 3 4 5 6; 1 3 6 10 15 21; 1 4 10 20 35 56; 1 5 15 35 70 126; 1 6 21 56 126 252]
```

```julia
x0=[1; 1; 1; 1; 1; 1]/sqrt(6)
```

```julia
norm(x0,2)
```

```julia
eigvals(A)
```

```julia
μ = RAY(A,x0,100)
```

```julia
function RAY_tol(A,n,eps)
    m = size(A,1)
    x = 100*rand(m)
    x = x/norm(x,2)
    μ = 0
    for i = 1:n
        λ = μ
        μ = (x'*A*x)/(x'*x)
        B = (A-μ*I)
        z = B\x
        x = z/norm(z,2)
        println("El valor de la iteración es = \n",μ)
        if abs(λ-μ)<eps
            return μ
        end
    end
    return μ
end
```

```julia
n = 5
B0 = 100*rand(n,n);
A0 = B0'*B0
```

```julia
eigvals(A0)
```

```julia
μ = RAY_tol(A0,100,10^(-12))
```

```julia
function RAY_cyc(A,eps)
    m = size(A,1)
    x = 100*rand(m)
    x = x/norm(x,2)
    μ = 0
    τ = 1
    while τ > eps
        λ = μ
        μ = (x'*A*x)/(x'*x)
        B = (A-μ*I)
        z = B\x
        x = z/norm(z,2)
        println("El valor de la iteración es = \n",μ)
        τ=abs(λ-μ)
    end
    return μ
end
```

```julia
μ = RAY_cyc(A0,10^(-2))
```
