# Iteraciones de Potencias

```julia
function ItQR(A,n)
    B= I
    T= A
    for i = 1:n
        Q1, R1 = qr(T)
        T = R1*Q1
    end
    return T
end
```

# Método de Potencias

Para $A \in \mathbb{C}^{n\times n} $ diagonalizable, tal que $X^{-1}AX = diag(\lambda_{1},\cdots,\lambda_{2})$ con $X=[x_{1},\cdots,x_{n}]$, y $|\lambda_{1}| > |\lambda_{2}| \geq \cdots \geq |\lambda_{n}|$. Sea $q^{(0)} \in \mathbb{C}^{n}$ un vector unitario en la norma $2$.

```julia
using Printf
```

```julia
function PM(A,q,n)
    p = q/norm(q,2)
    λ = 1
    for i = 1:n
        z = A*p
        p = z/norm(z,2)
        λ = p'*A*p
        println("Lambda = \n",λ)
    end
    return λ
end
```

```julia
F= [2 0 0;0 3 0;0 0 5]
```

```julia
eigvals(F)
```

```julia
α = PM(F,v,80)
```

# Iteración ortogonal

$\textbf{Teorema:}$ Sea la descomposición de Schur de $A \in \mathbb{C}^{n \times n}$ dada por $(1)$ y $(2)$ con $n \geq 2$. Asumamos que $|\lambda_{r}| > |\lambda_{r+1}|$ y que $\theta \geq 0$ satisface que

$$(1+\theta)|\lambda_{r}| > ||N||_{F}.$$

Si $Q_{0} \in \mathbf{C}^{n \times r}$ tiene columnas ortonormales y

$$d=dist(D_{r}(A^{H}),ran(Q_{0})) < 1,$$

entonces las matrices $Q_{k}$ generadas por el método de iteración ortogonal satisfacen que

$$dist(D_{r}(A),ran(Q_{k})) \leq \frac{(1+\theta)^{n-2}}{\sqrt{1-d^{2}}}\left(1+\frac{||T_{12}||_{F}}{sep(T_{11},T_{22})}\right) \left(\frac{|\lambda_{r+1}| + ||N||_{F}/(1+\theta)}{|\lambda_{r}|-||N||_{F}/(1+\theta)}\right)^{k}$$

```julia
function IO(A,Q,n)
    for i = 1:n
        Z = A*Q
        Q1, R1 = qr(Z)
        Q= Q1
    end
    return Q
end
```

# Iteraciones LR

```julia
L1, U1 = lu(F1)
```

Sean $A \in \mathbb{C}^{n \times n}$ y $G_{0} \in \mathbb{C}^{n \times r}$ con $rank(G_{0}) = r$. La iteración siguiente se conoce como $\textbf{treppeniteration}$.

```julia
function Trep(A,G,n)
    for i = 1:n
        Z = A*G
        L, U = lu(Z)
        G = L 
    end
    return Q
end
```

Si $r=n$ y bajo suposiciones razonables se puede deducir el siguiente método el cual se conoce como $\textbf{LR iteration}$.

```julia
function LR(A,G,n)
    T= inv(G)*A*G
    for i = 1:n
        L, U = lu(T)
        T = U*L
    end
    return T
end
```
