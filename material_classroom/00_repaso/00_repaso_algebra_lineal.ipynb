# Multiplicación de matricial

# Fila y columna
En esta parte tenemos dos casos, el caso en donde se multiplica un vector fila por un vector columna o el caso en que se multiplica un vector columna por un vetor fila.
1. Sea $v = (v_1 \quad v_2 \quad \dots \quad v_n)$ y sea $x = \begin{pmatrix} x_1 \\ x_2 \\ \vdots \\x_n \end{pmatrix}$
tenemos entonces que, 
$$
v \cdot x = v_1x_1 + v_2x_2 + \dots +v_nx_n.
$$
Este primer caso se le llama Producto interior.
2. Con los mismos vectores $x,v$ ahora consideremos el caso contrario es decir,
$$
x \cdot v = \begin{pmatrix}
x_1v1 & x_1v_2 & \dots & x_1v_n\\
x_2v1 & x_2v_2 & \dots & x_2v_n \\
\vdots & \vdots & \ddots & \vdots \\
x_nv1 & x_nv_2 & \dots & x_nv_n.
\end{pmatrix}
$$
A este caso se le llama Producto exterior.

Veamos ahora como hacer esto en Julia.

```julia
#Vector fila por vector columna
x = rand(5,1) 
y = rand(5,1)
# x' simboliza en Julia el vector transpuesto
xy = x'y
```

```julia
#Vector columna por vector fila
yx = y*x'
```

# Matriz por columna = columna

Podemos ver este caso $Ax$ de dos formas diferentes, como se presenta a continuación:

$$
Ax = \begin{pmatrix}
    a_{11} & a_{12} & \dots & a_{1n} \\
    a_{21} & a_{22} & \dots & a_{2n} \\
    \vdots & \vdots & \ddots & \vdots \\
    a_{m1} & a_{m2} & \dots & a_{mn}. 
        \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \\ \vdots \\x_n \end{pmatrix} = \begin{pmatrix} f_1(A)x_1 \\ f_2(A)x_2 \\ \vdots \\f_n(A)x_n \end{pmatrix}.
$$
Donde podemos ver que lo que se hace es tomar la fila $i$ y se multiplica por el vecto $x$ como veiamos en el primera caso de la multiplicación de fila por columna.

Para la segunda forma lo que hacemos es tomar las columans de $A$, es decir,
$$
Ax = \begin{pmatrix}
    a_{11} & a_{12} & \dots & a_{1n} \\
    a_{21} & a_{22} & \dots & a_{2n} \\
    \vdots & \vdots & \ddots & \vdots \\
    a_{m1} & a_{m2} & \dots & a_{mn}. 
        \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \\ \vdots \\x_n \end{pmatrix} = x_1C_1(A) + x_2C_2(A) + \dots + x_nC_n(A)
$$



```julia
#podemos tomar un vector x y una matriz A cualquiera y el producto de esta matriz con este vector.
A = rand(4,5)
x = rand(5,1)
b = A*x
```

## Fila x Matriz = Fila

Consideramos entonces un vector fila $(v_1 \quad v_2 \quad \dots \quad v_p)$ y una matriz $A$ de tamaño $p\times n$, tenemos entonces que 
$$
vA = \begin{pmatrix}v_1 & v_2 & \dots & v_p\end{pmatrix} \begin{pmatrix}
    a_{11} & a_{12} & \dots & a_{1n} \\
    a_{21} & a_{22} & \dots & a_{2n} \\
    \vdots & \vdots & \ddots & \vdots \\
    a_{p1} & a_{p2} & \dots & a_{pn}
        \end{pmatrix}.  
$$
Tenemos dos formas de hacer esta multiplicación, la primera es tomar $v$ y multiplicarlo por cada $C_i(A)$ dando como resultado que 
$$
vA = \begin{pmatrix} vC_1(A) & vC_2(A) & \dots vC_n(A)  \end{pmatrix},
$$
y la segunda forma que es ver la matriz $A$ como un vector columna cuyas entradas son las filas de $A$, es decir, 
$$
A = \begin{pmatrix} f_1(A) \\ f_2(A) \\ \vdots \\f_p(A) \end{pmatrix}.
$$
Entonces, 
$$
vA = \begin{pmatrix}v_1 & v_2 & \dots & v_p\end{pmatrix} \begin{pmatrix} f_1(A) \\ f_2(A) \\ \vdots \\f_n(A) \end{pmatrix} = v_1f_1(A) + v_2f_2(A) + \dots + v_pf_p(A)
$$

```julia
#Veamos este tipo de multiplicación en Julia
p = 5
v = rand(1, 5)
A = rand(5, 4)
v*A
```

# Sistemas lineales
Podemos ahora recordar un sistema lineal, tenemos entonces un sistema de $m$ ecuaciones con $n$ incógnitas, esto lo podemos representar matricialmente como 
$$
Ax = b
$$
donde $A$ es una matriz $m\times n$, $x$ es un vector de tamaño $n\times 1$ y $b$ es un vector de tamaño $m\times 1$. En un sistema lineal conocemos $a_{ij},b_i$ y buscamos encontrar $x_j$ con $i = 1, \dots m$ y $j = 1,\dots n$. Viendo las $m$ filas por separado y haciendo la respectiv multiplicación lo que buscamos solucionar es
$$
a_{i1}x_1 + a_{i2}x_2 + \dots + a_{in}x_n = b_i.
$$



## Matriz x Matriz = Matriz
Ahora podemos recordar la multiplicación en matrices, en donde se pueden presentar 4 formas diferentes de realizar esta operación. Para ellos consideremos dos matrices $A,B$ en donde $A$ sea de dimensión $m\times n$ y $B$ sea de dimensión $n\times p$.
### Forma 1
Para esta forma, lo que hacemos es ver a $B$ como un vector fila, cuyas entradas son sus columnas y $A$ lo dejamos quieto, entonces vamos a tener que

$$
AB = A\begin{pmatrix}C_1(B) & C_2(B) & \dots & C_p(B)\end{pmatrix} = \begin{pmatrix}AC_1(B) & AC_2(B) & \dots & AC_p(B)\end{pmatrix}.
$$
### Forma 2
Para esta forma, vemos a $A$ como un vector columna cuyas entradas son las filas de $A$ y a $B$ se deja quieto, vamos a tener que,
$$
AB = \begin{pmatrix} f_1(A) \\ f_2(A) \\ \vdots \\f_m(A) \end{pmatrix}B = \begin{pmatrix} f_1(A)B \\ f_2(A)B \\ \vdots \\f_n(A)B \end{pmatrix}.
$$
### Forma 3
Para este caso vamos a considerar a $A$ como un vector columna como en la forma 2 y vamos a considerar a $B$ como un vector fila como en la forma 1, de esta manera vamos a tener que
$$
AB = \begin{pmatrix} f_1(A) \\ f_2(A) \\ \vdots \\f_m(A) \end{pmatrix}\begin{pmatrix}C_1(B) & C_2(B) & \dots & C_p(B)\end{pmatrix} = \begin{pmatrix}
fl_1(A)C_1(B) & fl_1(A)C_2(B) & \dots & fl_1(A)C_p(B)\\
fl_2(A)C_1(B) & fl_2(A)C_2(B) & \dots & fl_2(A)C_p(B) \\
\vdots & \vdots & \ddots & \vdots \\
fl_m(A)C_1(B) & fl_m(A)C_2(B) & \dots & fl_m(A)C_p(B)
\end{pmatrix}.
$$
### Forma 4
Por último lo que hacemos es ver a $A$ como un vector fila cuyas entradas son las columnas de $A$ y a $B$ la vemos como un vector columna cuyas entradas son las filas de $B$, tenemos entonces que el producto es
$$
AB = \begin{pmatrix}C_1(A) & C_2(A) & \dots & C_n(A)\end{pmatrix} \begin{pmatrix} f_1(B) \\ f_2(B) \\ \vdots \\f_n(B) \end{pmatrix} = C_1(A)f_1(B) + C_2(A)f_2(B) + \dots + C_n(A)f_n(B).
$$
Y como es de esperarse una multiplicación de matrices en Julia se hace de la siguiente manera:
